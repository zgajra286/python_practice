```python
def create_age_group(age: int) -> str:
    if age < 18:
        return "Young"
    elif age >= 18 and age <=50:
        return "Middle Aged"
    else:
        return "Senior"

def data_preprocess(data_frame: pd.DataFrame) -> pd.DataFrame:
    df = data_frame.copy()

    df["diagnosed_diabetes"] = df["diagnosed_diabetes"].map(
        {
            1:"Positve", 0: "Negative"
        }
    )
    # Feature Engineering
    df["age_group"] = df["age"].apply(create_age_group)

    return df
```

```python
def plot_pie(column: str, figsize = (3,3)):
    data_ = data[column].value_counts()
    
    plt.figure(figsize = figsize)
    plt.pie(data_.values, labels = data_.index, autopct = '%1.2f%%')
    plt.legend(data_.index, loc = "upper right")
    plt.show()

def plot_hist(column: str, figsize = (3,3), xlabel = "", title: str = ""):
    plt.figure(figsize = figsize)
    plt.hist(data[column])
    plt.xlabel(xlabel) if xlabel else plt.xlabel(column)
    plt.title(title) if title else plt.title(column)
    plt.show()
```

```python
plot_pie("education_level", figsize = (5,5))
```
